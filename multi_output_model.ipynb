```python
import pandas as pd
from sklearn.datasets import load_linnerud
from sklearn.multioutput import MultiOutputRegressor
from sklearn.ensemble import GradientBoostingRegressor
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import accuracy_score, classification_report,mean_squared_error,r2_score
```

```python
line = load_linnerud(as_frame = True)
x = line.data
y = line.target
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size= 0.1,random_state =42)
rf_model = RandomForestRegressor(n_estimators=100, max_depth=3,random_state=42)
print("training in progress...")
rf_model.fit(x_train, y_train)
y_pred = rf_model.predict(x_test)
y_pred_df = pd.DataFrame(y_pred,columns = y.columns,index = y_test.index)


print("training completed")
for column in y.columns:
  mse = mean_squared_error(y_test[column],y_pred_df[column])
  r2 = r2_score(y_test[column],y_pred_df[column])
  print(f"Mean squared error for {column}: {mse}")
  print(f"R2 score for {column}: {r2}")
```
